```python
# 2. Visualize a vector gradient field
from field_runtime import Field

f = Field.preset("radial").gradient()
f
```

```python
# 3. Analytic mathematical expression with isosurface and planar slice
saddle = Field.expr("x**2 - y**2 + 0.5*z**2", domain={"x": (-4, 4), "y": (-4, 4), "z": (-4, 4)})
saddle.view.iso(level=1.0, color="#1D4ED8", opacity=0.9)
saddle.view.slice(axis="z", value=0.0, cmap="plasma")
saddle
```

```python
# 4. 3D Vortex Whirlpool with custom domain sampling
vortex = Field.vector("[-y, x, 0.2*z]")
vortex.sample(res=64, box={"x": (-5, 5), "y": (-5, 5), "z": (-5, 5)})
vortex
```
